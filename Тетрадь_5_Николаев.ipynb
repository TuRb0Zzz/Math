## Задание 5.1. Независимость остатков от деления и ковариация

**Условие:**
Дискретная случайная величина $X$ принимает только целые значения 1; 2; 3; 4; 5 и 6, при этом все указанные значения равновероятны. Пусть $Y_n$ – остаток от деления $X$ на $n$ ($n = 2$ или $n = 3$). Верно ли, что $Y_2$ и $Y_3$ независимы? Решить ту же задачу, если распределение $X$ пропорционально принимаемым значениям, т.е. $P(X = k) = A\cdot k, k = 1, \dots, 6$. Найти также ковариацию между $Y_2$ и $Y_3$.


```python
import numpy as np

vals = np.arange(1, 7)
p_unif = np.ones(6) / 6
y2 = vals % 2
y3 = vals % 3

e_y2_u = np.sum(y2 * p_unif)
e_y3_u = np.sum(y3 * p_unif)
e_y2y3_u = np.sum(y2 * y3 * p_unif)
cov_u = e_y2y3_u - e_y2_u * e_y3_u

p_grid_u = np.zeros((2, 3))
for i in range(6):
    p_grid_u[y2[i], y3[i]] += p_unif[i]
p_prod_u = np.outer(np.bincount(y2, p_unif), np.bincount(y3, p_unif))
is_indep_u = np.allclose(p_grid_u, p_prod_u)

p_prop = vals / np.sum(vals)
e_y2_p = np.sum(y2 * p_prop)
e_y3_p = np.sum(y3 * p_prop)
e_y2y3_p = np.sum(y2 * y3 * p_prop)
cov_p = e_y2y3_p - e_y2_p * e_y3_p

p_grid_p = np.zeros((2, 3))
for i in range(6):
    p_grid_p[y2[i], y3[i]] += p_prop[i]
p_y2_p = np.array([np.sum(p_prop[y2 == 0]), np.sum(p_prop[y2 == 1])])
p_y3_p = np.array([np.sum(p_prop[y3 == 0]), np.sum(p_prop[y3 == 1]), np.sum(p_prop[y3 == 2])])
p_prod_p = np.outer(p_y2_p, p_y3_p)
is_indep_p = np.allclose(p_grid_p, p_prod_p)

print(f"Равномерное распределение: независимы = {is_indep_u}, Cov(Y2, Y3) = {cov_u:.6f}")
print(f"Пропорциональное распределение: независимы = {is_indep_p}, Cov(Y2, Y3) = {cov_p:.6f}")
```

## Задание 5.2. Сумма независимых случайных величин

**Условие:**
Независимые случайные величины $X, Y, Z$ могут принимать только целые значения:
- $X$ – от 0 до 12 c вероятностью $\frac{1}{13}$,
- $Y$ – от 0 до 13 с вероятностью $\frac{1}{14}$,
- $Z$ – только значения 3 и 7, при этом $P(Z = 3) = \frac{9}{10}, P(Z = 7) = \frac{1}{10}$.

Найдите:
- a) вероятность того, что сумма данных случайных величин будет равна 12;
- б) наиболее вероятное значение суммы $X + Y + Z$ и вероятность такого события;
- в) распределение $X + Y + Z$.


```python
import numpy as np
import matplotlib.pyplot as plt

px = np.ones(13) / 13
py = np.ones(14) / 14
pz = np.zeros(8)
pz[3] = 0.9
pz[7] = 0.1

pxy = np.convolve(px, py)
pxyz = np.convolve(pxy, pz)

sums = np.arange(len(pxyz))
p_12 = pxyz[12]
mode_val = int(np.argmax(pxyz))
p_mode = pxyz[mode_val]

print(f"a) P(X + Y + Z = 12) = {p_12:.17f}")
print(f"б) M = {mode_val}, P(X + Y + Z = M) = {p_mode:.16f}")

fig, ax = plt.subplots(figsize=(11, 4.5))
ax.bar(sums, pxyz, color="cornflowerblue", edgecolor="navy", width=0.7)
ax.axvline(mode_val, color="red", linestyle="--", label=f"M = {mode_val}")
ax.set_title("Распределение суммы X + Y + Z")
ax.set_xlabel("Значение суммы")
ax.set_ylabel("Вероятность")
ax.set_xticks(sums[::2])
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

## Задание 5.3. Вероятность различных значений трех случайных величин

**Условие:**
Независимые случайные величины $X, Y, Z$ принимают только целые значения:
- $X$ – от 1 до 13 с вероятностью $P(X = i) = A\cdot i, i = 1, \dots, 13$;
- $Y$ – от 1 до 12 с вероятностью $P(Y = j) = B\cdot j, j = 1, \dots, 12$;
- $Z$ – от 1 до 8 с вероятностью $P(Z = k) = C\cdot k, k = 1, \dots, 8$.

Найдите вероятность того, что $X, Y, Z$ примут разные значения, т.е. $P(X \neq Y, X \neq Z, Y \neq Z)$.


```python
import numpy as np

x_vals = np.arange(1, 14)
y_vals = np.arange(1, 13)
z_vals = np.arange(1, 9)

px = x_vals / np.sum(x_vals)
py = y_vals / np.sum(y_vals)
pz = z_vals / np.sum(z_vals)

prob_distinct = 0.0
for i, x in enumerate(x_vals):
    for j, y in enumerate(y_vals):
        if x == y:
            continue
        for k, z in enumerate(z_vals):
            if x != z and y != z:
                prob_distinct += px[i] * py[j] * pz[k]

print(f"P(X != Y, X != Z, Y != Z) = {prob_distinct:.16f}")
```

## Задание 5.7. Линейная комбинация индикаторов попарно независимых событий

**Условие:**
События $A, B$ и $C$ имеют вероятности: $P(A) = 0{,}4, P(B) = 0{,}3, P(C) = 0{,}2$. Эти события попарно независимы, но все три одновременно наступить не могут. Пусть $X$ – индикатор $A$, $Y$ – индикатор $B$, $Z$ – индикатор $C$, а $V = 2X + 4Y + 6Z$. Найдите:
1) математическое ожидание $E(V)$;
2) дисперсию $\mathrm{Var}(V)$.


```python
pa, pb, pc = 0.4, 0.3, 0.2

e_v = 2 * pa + 4 * pb + 6 * pc
var_x = pa * (1 - pa)
var_y = pb * (1 - pb)
var_z = pc * (1 - pc)
var_v = (2**2) * var_x + (4**2) * var_y + (6**2) * var_z

print(f"1) E(V) = {e_v:.4f}")
print(f"2) Var(V) = {var_v:.4f}")
```

## Задание 5.9. Сумма и произведение индикаторов случайных точек

**Условие:**
Внутри квадрата площади 100 расположены треугольник и круг. Площади этих фигур даны: треугольник – 42, круг – 58. Также известно, что площадь пересечения треугольника и круга равна 24. В квадрате случайным независимым образом выбираются точки $\omega_1, \dots, \omega_5$. Определим случайные величины: $X_i$ – индикатор попадания $\omega_i$ в треугольник, $Y_i$ – индикатор попадания $\omega_i$ в круг, $Z_i = X_i + Y_i, i = 1, \dots, 5$. Определим также сумму $U = Z_1 + \dots + Z_5$ и произведение $V = Z_1 \dots Z_5$. Найдите:
1) математическое ожидание $E(U)$;
2) дисперсию $\mathrm{Var}(U)$;
3) математическое ожидание $E(V)$;
4) дисперсию $\mathrm{Var}(V)$.


```python
p_tri = 42 / 100
p_cir = 58 / 100
p_both = 24 / 100

p_z2 = p_both
p_z1 = (p_tri - p_both) + (p_cir - p_both)
p_z0 = 1 - p_z1 - p_z2

ez = 0 * p_z0 + 1 * p_z1 + 2 * p_z2
ez2 = (0**2) * p_z0 + (1**2) * p_z1 + (2**2) * p_z2
var_z = ez2 - ez**2

e_u = 5 * ez
var_u = 5 * var_z
e_v = ez**5
var_v = (ez2**5) - (e_v**2)

print(f"1) E(U) = {e_u:.3f}")
print(f"2) Var(U) = {var_u:.4f}")
print(f"3) E(V) = {e_v:.4f}")
print(f"4) Var(V) = {var_v:.4f}")
```

## Задание 5.10. Характеристики суммы независимых одинаково распределенных величин

**Условие:**
Случайные величины $X_i$, где $i = 1, 2, 3$, независимы и одинаково распределены:
$$
\begin{array}{c|cccc}
X_i & 2 & 5 & 8 & 10 \\
\hline
P & 0{,}25 & 0{,}25 & 0{,}25 & 0{,}25
\end{array}
$$
Пусть $S = X_1 + X_2 + X_3$. Найдите:
1) математическое ожидание $E(S)$;
2) наименьшее число $\mathrm{Med}_{\min}$, для которого $P(S \le \mathrm{Med}_{\min}) \ge 0{,}5$;
3) наибольшее число $\mathrm{Med}_{\max}$, для которого $P(S \ge \mathrm{Med}_{\max}) \ge 0{,}5$;
4) наименьшее число $\mathrm{Mod}_{\min}$, вероятность которого максимальна;
5) наибольшее число $\mathrm{Mod}_{\max}$, вероятность которого максимальна.


```python
import numpy as np

vals = np.array([2, 5, 8, 10])
e_s = 3 * np.mean(vals)

grid = np.array(np.meshgrid(vals, vals, vals)).T.reshape(-1, 3)
sums = np.sum(grid, axis=1)

unique_s, counts = np.unique(sums, return_counts=True)
p_s = counts / len(sums)

cdf_le = np.cumsum(p_s)
cdf_ge = np.cumsum(p_s[::-1])[::-1]

med_min = unique_s[cdf_le >= 0.5][0]
med_max = unique_s[cdf_ge >= 0.5][-1]

max_p = np.max(p_s)
modes = unique_s[p_s == max_p]
mod_min = np.min(modes)
mod_max = np.max(modes)

print(f"1) E(S) = {e_s:.3f}")
print(f"2) MedMin = {med_min}")
print(f"3) MedMax = {med_max}")
print(f"4) ModMin = {mod_min}")
print(f"5) ModMax = {mod_max}")
```

## Задание 5.11. Моделирование цены акции во времени

**Условие:**
Вероятность повышения цены акции на 4% равна 0,1; вероятность повышения на 0,3% равна 0,5; а вероятность понижения на 1% равна 0,4.
Пусть $S_{160}$ – цена акции после 160 рабочих дней. Предполагая, что начальная цена акции составляет $S_0 = 1000$ рублей, а относительные изменения цены за различные рабочие дни – независимые случайные величины, найдите:
1) математическое ожидание цены $S_{160}$;
2) стандартное отклонение цены $S_{160}$.


```python
import numpy as np

s0 = 1000
days = 160

r_vals = np.array([1 + 0.04, 1 + 0.003, 1 - 0.01])
p_r = np.array([0.1, 0.5, 0.4])

e_r = np.sum(r_vals * p_r)
e_r2 = np.sum((r_vals**2) * p_r)

e_s160 = s0 * (e_r ** days)
e_s160_sq = (s0**2) * (e_r2 ** days)
var_s160 = e_s160_sq - e_s160**2
std_s160 = np.sqrt(var_s160)

print(f"1) Математическое ожидание = {e_s160:.3f}")
print(f"2) Стандартное отклонение = {std_s160:.3f}")
```

## Задание 5.12. Портфель минимальной дисперсии из двух активов

**Условие:**
Математическое ожидание доходности акций компаний А и В составляет 4% и 5%, при этом стандартное отклонение доходности равно 5% и 8%, соответственно. Коэффициент корреляции $\rho_{AB} = 0{,}26$. Найдите (короткие позиции допускаются):
1) доли акций А и В в портфеле с минимальной дисперсией доходности;
2) ожидаемую доходность и стандартное отклонение доходности такого портфеля.


```python
import numpy as np

mu_a, mu_b = 4.0, 5.0
sigma_a, sigma_b = 5.0, 8.0
rho = 0.26
cov = rho * sigma_a * sigma_b

var_a = sigma_a**2
var_b = sigma_b**2

w_a = (var_b - cov) / (var_a + var_b - 2 * cov)
w_b = 1 - w_a

mu_p = w_a * mu_a + w_b * mu_b
var_p = (w_a**2) * var_a + (w_b**2) * var_b + 2 * w_a * w_b * cov
sigma_p = np.sqrt(var_p)

print(f"1) Доля А = {w_a:.3f}; доля В = {w_b:.3f}")
print(f"2) Ожидаемая доходность = {mu_p:.2f}%; стандартное отклонение = {sigma_p:.2f}%")
```

## Задание 5.13. Портфель минимальной дисперсии из трех независимых активов

**Условие:**
Ожидаемая доходность и стандартное отклонение доходности за период для акций компаний А, В, С составляют 1%, 3%, 3% и 2%, 6%, 5%, соответственно. Предполагая независимость доходностей акций А, В и С, найдите (в %) ожидаемую доходность портфеля, составленного из этих акций так, чтобы дисперсия его доходности была минимальной.


```python
import numpy as np

mu = np.array([1.0, 3.0, 3.0])
sigmas = np.array([2.0, 6.0, 5.0])
vars_ = sigmas**2

inv_vars = 1 / vars_
weights = inv_vars / np.sum(inv_vars)

exp_return = np.sum(weights * mu)

print(f"Ожидаемая доходность = {exp_return:.3f}%")
```

## Задание 5.14. Стандартное отклонение портфеля инвестора

**Условие:**
Инвестор сформировал портфель из акций компаний А и В, затратив на приобретение акций А в 7 раз больше средств чем на покупку акций В. Ожидаемая доходность за период владения акциями А и В составляет 3% и 2%, при этом стандартное отклонение доходности равно 4% и 4%, соответственно. Коэффициент корреляции доходностей акций А и В равен 0,3. Найдите (в %) стандартное отклонение доходности портфеля инвестора.


```python
import numpy as np

w_a = 7 / 8
w_b = 1 / 8
sig_a = 4.0
sig_b = 4.0
rho = 0.3
cov = rho * sig_a * sig_b

var_p = (w_a**2) * (sig_a**2) + (w_b**2) * (sig_b**2) + 2 * w_a * w_b * cov
std_p = np.sqrt(var_p)

print(f"Стандартное отклонение = {std_p:.3f}%")
```

## Задание 5.15. Характеристики суммы независимых пуассоновских величин

**Условие:**
Независимые пуассоновские случайные величины $X_1, X_2, X_3$ имеют следующие стандартные отклонения: $\sigma_1 = 0{,}8; \sigma_2 = 1{,}8; \sigma_3 = 0{,}4$. Пусть $X = X_1 + X_2 + X_3$. Найдите:
1) вероятность $P(X = 7)$;
2) наиболее вероятное значение суммы $X$;
3) стандартное отклонение $\sigma_X$;
4) асимметрию $\mathrm{As}(X)$;
5) эксцесс $\mathrm{Ex}(X)$.


```python
import numpy as np
from math import factorial

sigmas = [0.8, 1.8, 0.4]
lambdas = [s**2 for s in sigmas]
lam = sum(lambdas)

p_7 = (lam**7) * np.exp(-lam) / factorial(7)
mode_val = int(np.floor(lam))
std_val = np.sqrt(lam)
skewness = 1 / np.sqrt(lam)
kurtosis = 1 / lam

print(f"1) P(X = 7) = {p_7:.4f}")
print(f"2) Наиболее вероятное значение = {mode_val}")
print(f"3) Стандартное отклонение = {std_val:.4f}")
print(f"4) Асимметрия = {skewness:.4f}")
print(f"5) Эксцесс = {kurtosis:.4f}")
```
